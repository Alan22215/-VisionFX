```python
import cv2
import os
import time

# --------------------------------------------------
# VisionFX R&D - Step 5D
# Basic Frame Processing Pipeline
# --------------------------------------------------

input_path = "/content/sa.mp4"
output_path = "/content/visionfx_test_output.mp4"

cap = cv2.VideoCapture(input_path)

if not cap.isOpened():
    raise RuntimeError("❌ Could not open input video.")

# Video properties
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

# MP4 output
fourcc = cv2.VideoWriter_fourcc(*"mp4v")

out = cv2.VideoWriter(
    output_path,
    fourcc,
    fps,
    (width, height)
)

print("========== VisionFX Processing ==========")
print("Input resolution :", f"{width} × {height}")
print("Input FPS        :", fps)
print("Total frames     :", total_frames)
print("-----------------------------------------")

start_time = time.time()
processed_frames = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # --------------------------------------------------
    # VisionFX PROCESSING AREA
    #
    # Future:
    # AI Detection
    # Segmentation
    # Tracking
    # Effects
    # GPU Processing
    # --------------------------------------------------

    processed_frame = frame

    # Write processed frame
    out.write(processed_frame)

    processed_frames += 1

cap.release()
out.release()

elapsed = time.time() - start_time

processing_fps = (
    processed_frames / elapsed
    if elapsed > 0 else 0
)

print("Frames processed :", processed_frames)
print("Processing time  :", round(elapsed, 3), "seconds")
print("Processing FPS   :", round(processing_fps, 2))
print("-----------------------------------------")
print("Output saved to   :", output_path)
print("=========================================")
```

```python
import cv2
import numpy as np
import time

# --------------------------------------------------
# VisionFX R&D v0.0.3
# Basic Neon Edge Effect
# --------------------------------------------------

input_path = "/content/sa.mp4"
output_path = "/content/visionfx_neon_test.mp4"

cap = cv2.VideoCapture(input_path)

if not cap.isOpened():
    raise RuntimeError("❌ Could not open input video.")

# Video properties
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

fourcc = cv2.VideoWriter_fourcc(*"mp4v")

out = cv2.VideoWriter(
    output_path,
    fourcc,
    fps,
    (width, height)
)

start_time = time.time()
processed_frames = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # ---------------------------------------------
    # 1. Convert frame to grayscale
    # ---------------------------------------------

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # ---------------------------------------------
    # 2. Detect edges
    # ---------------------------------------------

    edges = cv2.Canny(
        gray,
        threshold1=80,
        threshold2=160
    )

    # ---------------------------------------------
    # 3. Smooth the edges
    # ---------------------------------------------

    edges_blur = cv2.GaussianBlur(
        edges,
        (0, 0),
        sigmaX=3
    )

    # ---------------------------------------------
    # 4. Create glow
    # ---------------------------------------------

    glow = cv2.cvtColor(
        edges_blur,
        cv2.COLOR_GRAY2BGR
    )

    # ---------------------------------------------
    # 5. Create colored neon layer
    # ---------------------------------------------

    neon = np.zeros_like(frame)

    # Cyan-style neon
    neon[:, :, 0] = glow[:, :, 0]
    neon[:, :, 1] = glow[:, :, 0]

    # ---------------------------------------------
    # 6. Combine original + neon
    # ---------------------------------------------

    processed_frame = cv2.addWeighted(
        frame,
        0.65,
        neon,
        1.8,
        0
    )

    # ---------------------------------------------
    # 7. Add sharper edge layer
    # ---------------------------------------------

    edge_color = np.zeros_like(frame)

    edge_color[:, :, 0] = edges
    edge_color[:, :, 1] = edges

    processed_frame = cv2.addWeighted(
        processed_frame,
        1.0,
        edge_color,
        0.8,
        0
    )

    # ---------------------------------------------
    # Write frame
    # ---------------------------------------------

    out.write(processed_frame)

    processed_frames += 1

cap.release()
out.release()

elapsed = time.time() - start_time

processing_fps = (
    processed_frames / elapsed
    if elapsed > 0 else 0
)

print("========== VisionFX Neon Test ==========")
print("Frames processed :", processed_frames)
print("Processing time  :", round(elapsed, 3), "seconds")
print("Processing FPS   :", round(processing_fps, 2))
print("Output file      :", output_path)
print("=========================================")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# VisionFX R&D v0.0.5
# AI Smart Outline - Single Frame
# --------------------------------------------------

# Make sure we have a segmentation mask
if result.masks is None:
    raise RuntimeError("❌ No segmentation mask available.")

# Get the first person mask
mask = result.masks.data[0].cpu().numpy()

# Convert to binary mask
mask_uint8 = (mask * 255).astype(np.uint8)

# Resize to original frame resolution
mask_resized = cv2.resize(
    mask_uint8,
    (frame.shape[1], frame.shape[0]),
    interpolation=cv2.INTER_LINEAR
)

# Threshold the mask
_, binary_mask = cv2.threshold(
    mask_resized,
    127,
    255,
    cv2.THRESH_BINARY
)

# --------------------------------------------------
# Find OUTER boundary of the person
# --------------------------------------------------

contours, _ = cv2.findContours(
    binary_mask,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

# Create empty outline layer
outline = np.zeros_like(frame)

# Draw the outer silhouette
cv2.drawContours(
    outline,
    contours,
    -1,
    (255, 255, 0),   # Cyan
    thickness=5
)

# --------------------------------------------------
# Create glow
# --------------------------------------------------

glow_1 = cv2.GaussianBlur(
    outline,
    (0, 0),
    sigmaX=8
)

glow_2 = cv2.GaussianBlur(
    outline,
    (0, 0),
    sigmaX=20
)

# --------------------------------------------------
# Composite
# --------------------------------------------------

result_frame = frame.copy()

# Add large glow
result_frame = cv2.addWeighted(
    result_frame,
    1.0,
    glow_2,
    0.8,
    0
)

# Add medium glow
result_frame = cv2.addWeighted(
    result_frame,
    1.0,
    glow_1,
    1.2,
    0
)

# Add sharp outline
result_frame = cv2.addWeighted(
    result_frame,
    1.0,
    outline,
    1.5,
    0
)

# --------------------------------------------------
# Display
# --------------------------------------------------

frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
mask_rgb = cv2.cvtColor(binary_mask, cv2.COLOR_GRAY2RGB)
outline_rgb = cv2.cvtColor(result_frame, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(15, 6))

plt.subplot(1, 3, 1)
plt.imshow(frame_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(mask_rgb)
plt.title("AI Person Mask")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(outline_rgb)
plt.title("VisionFX Smart Outline")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from google.colab.output import eval_js
from IPython.display import display, Image
import base64
import cv2
import numpy as np

# --------------------------------------------------
# VisionFX R&D
# Step 8B - Direct Camera Capture
# --------------------------------------------------

js_code = """
async function captureCameraFrame() {

    const stream = await navigator.mediaDevices.getUserMedia({
        video: {
            facingMode: "user"
        },
        audio: false
    });

    const video = document.createElement("video");

    video.srcObject = stream;
    video.setAttribute("playsinline", "");
    video.muted = true;

    await video.play();

    // Wait until camera has a valid frame
    await new Promise(resolve => {
        if (video.readyState >= 2) {
            resolve();
        } else {
            video.onloadeddata = resolve;
        }
    });

    // Small delay so camera exposure/focus can settle
    await new Promise(resolve => setTimeout(resolve, 1000));

    const canvas = document.createElement("canvas");

    canvas.width = video.videoWidth;
    canvas.height = video.videoHeight;

    const context = canvas.getContext("2d");

    context.drawImage(
        video,
        0,
        0,
        canvas.width,
        canvas.height
    );

    // Stop camera immediately after capturing
    stream.getTracks().forEach(track => track.stop());

    return canvas.toDataURL("image/jpeg", 0.90);
}

captureCameraFrame();
"""

data = eval_js(js_code)

# --------------------------------------------------
# Convert browser image → OpenCV frame
# --------------------------------------------------

image_bytes = base64.b64decode(
    data.split(",")[1]
)

image_array = np.frombuffer(
    image_bytes,
    dtype=np.uint8
)

camera_frame = cv2.imdecode(
    image_array,
    cv2.IMREAD_COLOR
)

if camera_frame is None:
    raise RuntimeError("❌ Could not decode camera image.")

print("✅ Camera frame captured")
print("Frame shape:", camera_frame.shape)

# Display frame
display(Image(data=image_bytes))
```

```python
import time
import base64
import cv2
import numpy as np
import matplotlib.pyplot as plt

from google.colab.output import eval_js
from IPython.display import display, Javascript

# --------------------------------------------------
# VisionFX R&D
# Step 8F - Continuous Camera Prototype
# --------------------------------------------------

def capture_camera_frame():

    js_code = """
    async function captureFrame() {

        const stream = await navigator.mediaDevices.getUserMedia({
            video: {
                facingMode: "user",
                width: {ideal: 640},
                height: {ideal: 480}
            },
            audio: false
        });

        const video = document.createElement("video");

        video.srcObject = stream;
        video.setAttribute("playsinline", "");
        video.muted = true;

        await video.play();

        await new Promise(resolve => {
            if (video.readyState >= 2) {
                resolve();
            } else {
                video.onloadeddata = resolve;
            }
        });

        await new Promise(resolve => setTimeout(resolve, 300));

        const canvas = document.createElement("canvas");

        canvas.width = video.videoWidth;
        canvas.height = video.videoHeight;

        const ctx = canvas.getContext("2d");

        ctx.drawImage(
            video,
            0,
            0,
            canvas.width,
            canvas.height
        );

        stream.getTracks().forEach(
            track => track.stop()
        );

        return canvas.toDataURL(
            "image/jpeg",
            0.85
        );
    }

    captureFrame();
    """

    data = eval_js(js_code)

    image_bytes = base64.b64decode(
        data.split(",")[1]
    )

    image_array = np.frombuffer(
        image_bytes,
        dtype=np.uint8
    )

    frame = cv2.imdecode(
        image_array,
        cv2.IMREAD_COLOR
    )

    return frame


def apply_smart_neon(frame):

    results = model(
        frame,
        device=0,
        verbose=False
    )

    result = results[0]

    if result.masks is None:
        return frame

    # Find person
    person_index = None

    for i, box in enumerate(result.boxes):

        class_id = int(box.cls[0])
        class_name = model.names[class_id]

        if class_name == "person":
            person_index = i
            break

    if person_index is None:
        return frame

    # Person mask
    mask = result.masks.data[
        person_index
    ].cpu().numpy()

    mask_uint8 = (
        mask * 255
    ).astype(np.uint8)

    mask_resized = cv2.resize(
        mask_uint8,
        (frame.shape[1], frame.shape[0]),
        interpolation=cv2.INTER_LINEAR
    )

    _, binary_mask = cv2.threshold(
        mask_resized,
        127,
        255,
        cv2.THRESH_BINARY
    )

    # Outer contour
    contours, _ = cv2.findContours(
        binary_mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    outline = np.zeros_like(frame)

    cv2.drawContours(
        outline,
        contours,
        -1,
        (255, 255, 0),
        thickness=4
    )

    # Glow
    glow_large = cv2.GaussianBlur(
        outline,
        (0, 0),
        18
    )

    glow_medium = cv2.GaussianBlur(
        outline,
        (0, 0),
        7
    )

    # Composite
    output = cv2.addWeighted(
        frame,
        1.0,
        glow_large,
        0.8,
        0
    )

    output = cv2.addWeighted(
        output,
        1.0,
        glow_medium,
        1.2,
        0
    )

    output = cv2.addWeighted(
        output,
        1.0,
        outline,
        1.5,
        0
    )

    return output


# --------------------------------------------------
# Process 10 camera frames
# --------------------------------------------------

frames_to_process = 10

times = []

print("Starting VisionFX camera experiment...\n")

for i in range(frames_to_process):

    start = time.perf_counter()

    frame = capture_camera_frame()

    output = apply_smart_neon(frame)

    elapsed = time.perf_counter() - start

    times.append(elapsed)

    print(
        f"Frame {i + 1:02d} | "
        f"Processing time: {elapsed * 1000:.1f} ms | "
        f"FPS: {1 / elapsed:.2f}"
    )

print("\n===================================")
print("VisionFX Camera Experiment")
print("Frames processed:", frames_to_process)
print(
    "Average FPS:",
    round(1 / np.mean(times), 2)
)
print("===================================")

# Display final processed frame
output_rgb = cv2.cvtColor(
    output,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(8, 6))
plt.imshow(output_rgb)
plt.title("VisionFX Live Camera Prototype")
plt.axis("off")
plt.show()
```

```python
import base64
import cv2
import numpy as np
import time
import json

from google.colab.output import eval_js
from IPython.display import display, Image

# --------------------------------------------------
# VisionFX R&D
# Step 8G - Batched Camera Capture
# --------------------------------------------------

NUM_FRAMES = 10

print("Opening camera...")

js_code = f"""
async function captureFrames() {{

    const stream = await navigator.mediaDevices.getUserMedia({{
        video: {{
            width: {{ideal: 640}},
            height: {{ideal: 480}},
            facingMode: "user"
        }},
        audio: false
    }});

    const video = document.createElement("video");

    video.srcObject = stream;
    video.setAttribute("playsinline", "");
    video.muted = true;

    await video.play();

    await new Promise(resolve => {{
        if (video.readyState >= 2) {{
            resolve();
        }} else {{
            video.onloadeddata = resolve;
        }}
    }});

    const canvas = document.createElement("canvas");

    canvas.width = video.videoWidth;
    canvas.height = video.videoHeight;

    const ctx = canvas.getContext("2d");

    const frames = [];

    // Capture frames while camera stays open
    for (let i = 0; i < {NUM_FRAMES}; i++) {{

        ctx.drawImage(
            video,
            0,
            0,
            canvas.width,
            canvas.height
        );

        frames.push(
            canvas.toDataURL(
                "image/jpeg",
                0.75
            )
        );

        // Small interval between frames
        await new Promise(
            resolve => setTimeout(resolve, 100)
        );
    }}

    // Stop camera
    stream.getTracks().forEach(
        track => track.stop()
    );

    return frames;
}}

captureFrames();
"""

start_capture = time.perf_counter()

frames_data = eval_js(js_code)

capture_time = time.perf_counter() - start_capture

print(
    f"Camera capture completed in "
    f"{capture_time:.2f} seconds"
)

print(
    "Frames captured:",
    len(frames_data)
)

# --------------------------------------------------
# Convert frames to OpenCV
# --------------------------------------------------

camera_frames = []

for data in frames_data:

    image_bytes = base64.b64decode(
        data.split(",")[1]
    )

    image_array = np.frombuffer(
        image_bytes,
        dtype=np.uint8
    )

    frame = cv2.imdecode(
        image_array,
        cv2.IMREAD_COLOR
    )

    if frame is not None:
        camera_frames.append(frame)

print(
    "Frames decoded:",
    len(camera_frames)
)

if len(camera_frames) > 0:

    print(
        "Frame resolution:",
        camera_frames[0].shape
    )

    # Show first captured frame
    display(
        Image(
            data=base64.b64decode(
                frames_data[0].split(",")[1]
            )
        )
    )
```

```python
# --------------------------------------------------
# VisionFX R&D
# Step 8J - Pipeline Stage Profiler
# --------------------------------------------------

import cv2
import numpy as np
import time
import torch

stage_times = {
    "AI inference": [],
    "Mask transfer": [],
    "Mask resize": [],
    "Threshold": [],
    "Contour": [],
    "Outline": [],
    "Glow": [],
    "Composite": []
}

print("Starting VisionFX stage profiling...\n")

for frame in camera_frames:

    # ==============================================
    # 1. AI INFERENCE
    # ==============================================

    start = time.perf_counter()

    results = model(
        frame,
        device=0,
        verbose=False
    )

    torch.cuda.synchronize()

    stage_times["AI inference"].append(
        time.perf_counter() - start
    )

    result = results[0]

    # ==============================================
    # Find person
    # ==============================================

    person_index = None

    if result.boxes is not None:

        for j, box in enumerate(result.boxes):

            class_id = int(box.cls[0])
            class_name = model.names[class_id]

            if class_name == "person":
                person_index = j
                break

    if (
        person_index is None
        or result.masks is None
    ):
        continue

    # ==============================================
    # 2. MASK TRANSFER
    # ==============================================

    start = time.perf_counter()

    mask = result.masks.data[
        person_index
    ].cpu().numpy()

    stage_times["Mask transfer"].append(
        time.perf_counter() - start
    )

    # ==============================================
    # 3. MASK RESIZE
    # ==============================================

    start = time.perf_counter()

    mask_uint8 = (
        mask * 255
    ).astype(np.uint8)

    mask_resized = cv2.resize(
        mask_uint8,
        (
            frame.shape[1],
            frame.shape[0]
        ),
        interpolation=cv2.INTER_LINEAR
    )

    stage_times["Mask resize"].append(
        time.perf_counter() - start
    )

    # ==============================================
    # 4. THRESHOLD
    # ==============================================

    start = time.perf_counter()

    _, binary_mask = cv2.threshold(
        mask_resized,
        127,
        255,
        cv2.THRESH_BINARY
    )

    stage_times["Threshold"].append(
        time.perf_counter() - start
    )

    # ==============================================
    # 5. CONTOUR
    # ==============================================

    start = time.perf_counter()

    contours, _ = cv2.findContours(
        binary_mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    stage_times["Contour"].append(
        time.perf_counter() - start
    )

    # ==============================================
    # 6. OUTLINE
    # ==============================================

    start = time.perf_counter()

    outline = np.zeros_like(frame)

    cv2.drawContours(
        outline,
        contours,
        -1,
        (255, 255, 0),
        thickness=4
    )

    stage_times["Outline"].append(
        time.perf_counter() - start
    )

    # ==============================================
    # 7. GLOW
    # ==============================================

    start = time.perf_counter()

    glow_large = cv2.GaussianBlur(
        outline,
        (0, 0),
        18
    )

    glow_medium = cv2.GaussianBlur(
        outline,
        (0, 0),
        7
    )

    stage_times["Glow"].append(
        time.perf_counter() - start
    )

    # ==============================================
    # 8. COMPOSITE
    # ==============================================

    start = time.perf_counter()

    output = cv2.addWeighted(
        frame,
        1.0,
        glow_large,
        0.8,
        0
    )

    output = cv2.addWeighted(
        output,
        1.0,
        glow_medium,
        1.2,
        0
    )

    output = cv2.addWeighted(
        output,
        1.0,
        outline,
        1.5,
        0
    )

    stage_times["Composite"].append(
        time.perf_counter() - start
    )


# ==============================================
# RESULTS
# ==============================================

print("==============================================")
print("VISIONFX PIPELINE PROFILING")
print("==============================================")

total = 0

for stage, values in stage_times.items():

    if len(values) == 0:
        continue

    average_ms = np.mean(values) * 1000

    total += average_ms

    print(
        f"{stage:<20} : "
        f"{average_ms:>8.2f} ms"
    )

print("----------------------------------------------")
print(
    f"{'Measured stages':<20} : "
    f"{total:>8.2f} ms"
)

print(
    f"{'Estimated FPS':<20} : "
    f"{1000 / total:>8.2f}"
)

print("==============================================")
```

```python
# --------------------------------------------------
# VisionFX R&D
# Step 8L - Optimized Full Pipeline
# --------------------------------------------------

import cv2
import numpy as np
import time
import torch

optimized_times = []
optimized_frames = []

print("Starting optimized VisionFX pipeline...\n")

for i, frame in enumerate(camera_frames):

    start = time.perf_counter()

    # ==============================================
    # 1. AI SEGMENTATION
    # ==============================================

    results = model(
        frame,
        device=0,
        verbose=False
    )

    torch.cuda.synchronize()

    result = results[0]

    # ==============================================
    # 2. FIND PERSON
    # ==============================================

    person_index = None

    if result.boxes is not None:

        for j, box in enumerate(result.boxes):

            class_id = int(box.cls[0])
            class_name = model.names[class_id]

            if class_name == "person":
                person_index = j
                break

    output = frame.copy()

    # ==============================================
    # 3. SMART OUTLINE
    # ==============================================

    if (
        person_index is not None
        and result.masks is not None
    ):

        mask = result.masks.data[
            person_index
        ].cpu().numpy()

        mask_uint8 = (
            mask * 255
        ).astype(np.uint8)

        mask_resized = cv2.resize(
            mask_uint8,
            (
                frame.shape[1],
                frame.shape[0]
            ),
            interpolation=cv2.INTER_LINEAR
        )

        _, binary_mask = cv2.threshold(
            mask_resized,
            127,
            255,
            cv2.THRESH_BINARY
        )

        contours, _ = cv2.findContours(
            binary_mask,
            cv2.RETR_EXTERNAL,
            cv2.CHAIN_APPROX_SIMPLE
        )

        outline = np.zeros_like(frame)

        cv2.drawContours(
            outline,
            contours,
            -1,
            (255, 255, 0),
            thickness=4
        )

        # ==========================================
        # 4. DOWNSAMPLED GLOW
        # ==========================================

        height, width = outline.shape[:2]

        small_width = width // 2
        small_height = height // 2

        small_outline = cv2.resize(
            outline,
            (
                small_width,
                small_height
            ),
            interpolation=cv2.INTER_AREA
        )

        small_glow = cv2.GaussianBlur(
            small_outline,
            (0, 0),
            7
        )

        large_glow = cv2.resize(
            small_glow,
            (
                width,
                height
            ),
            interpolation=cv2.INTER_LINEAR
        )

        # ==========================================
        # 5. COMPOSITE
        # ==========================================

        output = cv2.addWeighted(
            frame,
            1.0,
            large_glow,
            1.2,
            0
        )

        output = cv2.addWeighted(
            output,
            1.0,
            outline,
            1.5,
            0
        )

    torch.cuda.synchronize()

    elapsed = time.perf_counter() - start

    optimized_times.append(elapsed)
    optimized_frames.append(output)

    print(
        f"Frame {i + 1:02d} | "
        f"{elapsed * 1000:.2f} ms | "
        f"{1 / elapsed:.2f} FPS"
    )

# ==============================================
# FINAL RESULTS
# ==============================================

average_time = np.mean(
    optimized_times
)

print("\n======================================")
print("VISIONFX OPTIMIZED PIPELINE")
print("======================================")

print(
    "Frames processed :",
    len(optimized_frames)
)

print(
    "Average time     :",
    round(average_time * 1000, 2),
    "ms"
)

print(
    "Estimated FPS    :",
    round(1 / average_time, 2)
)

print("======================================")
```

```python
# --------------------------------------------------
# VisionFX R&D
# Step 8M - Stable Warmed-Up Benchmark
# --------------------------------------------------

import cv2
import numpy as np
import time
import torch

print("Preparing VisionFX benchmark...")

# --------------------------------------------------
# Pipeline function
# --------------------------------------------------

def visionfx_process(frame):

    results = model(
        frame,
        device=0,
        verbose=False
    )

    result = results[0]

    person_index = None

    if result.boxes is not None:

        for j, box in enumerate(result.boxes):

            class_id = int(box.cls[0])
            class_name = model.names[class_id]

            if class_name == "person":
                person_index = j
                break

    output = frame.copy()

    if (
        person_index is not None
        and result.masks is not None
    ):

        # -------------------------------
        # Person mask
        # -------------------------------

        mask = result.masks.data[
            person_index
        ].cpu().numpy()

        mask_uint8 = (
            mask * 255
        ).astype(np.uint8)

        mask_resized = cv2.resize(
            mask_uint8,
            (
                frame.shape[1],
                frame.shape[0]
            ),
            interpolation=cv2.INTER_LINEAR
        )

        _, binary_mask = cv2.threshold(
            mask_resized,
            127,
            255,
            cv2.THRESH_BINARY
        )

        # -------------------------------
        # Outer contour
        # -------------------------------

        contours, _ = cv2.findContours(
            binary_mask,
            cv2.RETR_EXTERNAL,
            cv2.CHAIN_APPROX_SIMPLE
        )

        outline = np.zeros_like(frame)

        cv2.drawContours(
            outline,
            contours,
            -1,
            (255, 255, 0),
            thickness=4
        )

        # -------------------------------
        # Downsampled glow
        # -------------------------------

        height, width = outline.shape[:2]

        small_outline = cv2.resize(
            outline,
            (
                width // 2,
                height // 2
            ),
            interpolation=cv2.INTER_AREA
        )

        small_glow = cv2.GaussianBlur(
            small_outline,
            (0, 0),
            7
        )

        large_glow = cv2.resize(
            small_glow,
            (
                width,
                height
            ),
            interpolation=cv2.INTER_LINEAR
        )

        # -------------------------------
        # Composite
        # -------------------------------

        output = cv2.addWeighted(
            frame,
            1.0,
            large_glow,
            1.2,
            0
        )

        output = cv2.addWeighted(
            output,
            1.0,
            outline,
            1.5,
            0
        )

    return output


# --------------------------------------------------
# GPU / model warm-up
# --------------------------------------------------

print("Warming up VisionFX...")

for i in range(5):

    frame = camera_frames[i % len(camera_frames)]

    _ = visionfx_process(frame)

    torch.cuda.synchronize()

print("✅ Warm-up complete")


# --------------------------------------------------
# Stable benchmark
# --------------------------------------------------

benchmark_times = []

iterations = 30

print(
    f"\nRunning {iterations} benchmark iterations...\n"
)

for i in range(iterations):

    frame = camera_frames[
        i % len(camera_frames)
    ]

    start = time.perf_counter()

    _ = visionfx_process(frame)

    torch.cuda.synchronize()

    elapsed = time.perf_counter() - start

    benchmark_times.append(
        elapsed * 1000
    )

    print(
        f"{i + 1:02d}: "
        f"{elapsed * 1000:.2f} ms"
    )


# --------------------------------------------------
# Statistics
# --------------------------------------------------

times = np.array(benchmark_times)

average_ms = np.mean(times)
median_ms = np.median(times)
minimum_ms = np.min(times)
maximum_ms = np.max(times)

print("\n==========================================")
print("VISIONFX STABLE BENCHMARK")
print("==========================================")

print(
    "Iterations       :",
    iterations
)

print(
    "Average          :",
    round(average_ms, 2),
    "ms"
)

print(
    "Median           :",
    round(median_ms, 2),
    "ms"
)

print(
    "Minimum          :",
    round(minimum_ms, 2),
    "ms"
)

print(
    "Maximum          :",
    round(maximum_ms, 2),
    "ms"
)

print(
    "Average FPS      :",
    round(1000 / average_ms, 2)
)

print(
    "Median FPS       :",
    round(1000 / median_ms, 2)
)

print("==========================================")
```
